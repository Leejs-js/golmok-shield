```python
import pandas as pd
import os

# 1. 현재 주피터 노트북 파일이 있는 'project/notebook' 폴더 주소를 가져옵니다.
notebook_dir = os.getcwd()

# 2. 무조건 한 단계 위로 올라갔다('..')가 'raw' 폴더로 들어가는 절대 경로를 완벽하게 빌드합니다.
path_2023 = os.path.abspath(os.path.join(notebook_dir, '..', 'raw', 'data_2023.csv'))
path_2024 = os.path.abspath(os.path.join(notebook_dir, '..', 'raw', 'data_2024.csv'))
path_2025 = os.path.abspath(os.path.join(notebook_dir, '..', 'raw', 'data_2025.csv'))

# 3. 데이터 불러오기
df_2023 = pd.read_csv(path_2023, encoding='utf-8')
df_2024 = pd.read_csv(path_2024, encoding='utf-8')
df_2025 = pd.read_csv(path_2025, encoding='utf-8')

# 데이터 개수 출력
print(f"2023년 데이터 개수: {len(df_2023)}개")
print(f"2024년 데이터 개수: {len(df_2024)}개")
print(f"2025년 데이터 개수: {len(df_2025)}개")
```

```python
# 2025년의 영문 헤더명을 2023년 한글 헤더명으로 덮어씌우기
df_2025.columns = df_2023.columns

# 세 년도 데이터를 하나로 병합
merged_df = pd.concat([df_2023, df_2024, df_2025], ignore_index=True)

print(f"✅ 총 3개년 병합 데이터 개수: {len(merged_df)}개")
display(merged_df.head(3))
```

```python
# 저장 경로도 파이썬이 알아서 raw 폴더 안에 넣도록 지정합니다.
if current_dir.endswith('notebook'):
    save_path = os.path.join(current_dir, '..', 'raw', '서울시_상권분석_점포_3개년_합본.csv')
elif current_dir.endswith('raw'):
    save_path = '서울시_상권분석_점포_3개년_합본.csv'
else:
    save_path = os.path.join(current_dir, 'raw', '서울시_상권분석_점포_3개년_합본.csv')

merged_df.to_csv(save_path, index=False, encoding='utf-8-sig')
print("🎉 드디어 병합 파일 저장 완료!")
```
