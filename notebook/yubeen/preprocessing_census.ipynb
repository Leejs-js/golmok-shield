```python
#3개년 데이터 합치기 위해 칼럼 같은지 확인
print(local_census_23.columns.equals(local_census_24.columns))
print(local_census_23.columns.equals(local_census_25.columns))
```

```python
#3개년 데이터 합침

all_local_census = pd.concat(
    [local_census_23, local_census_24, local_census_25],
    axis=0,
    ignore_index=True
)

print(local_census_23.shape)
print(local_census_24.shape)
print(local_census_25.shape)
print(all_local_census.shape)
```

```python
#결측치 확인

from IPython.display import display

display(all_local_census.isna().sum().to_frame("NaN count"))
```

```python
area = area[area["시군구명"] == "마포구"]
area["행정동코드"] = area["행정동코드"] // 100
area
```

```python
# '행정동코드' 칼럼을 기준으로 조인


all_local_census = all_local_census.merge(
    area[["행정동코드", "읍면동명"]],
    on="행정동코드",
    how="left"
)

all_local_census.head()
```

```python
all_local_census = all_local_census[all_local_census["읍면동명"].notna()]

all_local_census.head()
```
