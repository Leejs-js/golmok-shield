```python
# =========================
# 마포구 행정동 KMeans 군집화 - One Cell Version
# =========================

# 필요하면 최초 1회만 실행
# !pip install pandas numpy scikit-learn matplotlib joblib

from pathlib import Path
import json
import joblib
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score, davies_bouldin_score

warnings.filterwarnings("ignore")

# =========================
# 1. 경로 설정
# =========================

DATA_PATH = Path("../../data/preprocessed/merged_datamart.csv")
OUTPUT_DIR = Path("../../outputs/clustering")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

K = 5  # 노션 기준: 대학가형 / 오피스형 / 주거형 / 관광형

# =========================
# 2. 데이터 불러오기
# =========================

df = pd.read_csv(DATA_PATH)

print("데이터 크기:", df.shape)
print("컬럼 수:", len(df.columns))
display(df.head())

# =========================
# 3. 유틸 함수
# =========================


def safe_divide(numerator, denominator):
    denominator = denominator.replace(0, np.nan)
    result = numerator / denominator
    return result.replace([np.inf, -np.inf], np.nan).fillna(0)


def minmax_series(s):
    if s.max() == s.min():
        return pd.Series([0.5] * len(s), index=s.index)
    return (s - s.min()) / (s.max() - s.min())


# =========================
# 4. 파생변수 생성
# =========================

df_model = df.copy()

# 인구 특성
df_model["young_pop_ratio"] = safe_divide(
    df_model["pop_20s"] + df_model["pop_30s"], df_model["tot_pop"]
)

df_model["middle_old_pop_ratio"] = safe_divide(
    df_model["pop_40s"] + df_model["pop_50s"] + df_model["pop_60s_above"],
    df_model["tot_pop"],
)

# 직장인 / 오피스 성격
df_model["office_worker_ratio"] = safe_divide(df_model["emp_cnt"], df_model["tot_pop"])

# 업종 비율
df_model["food_biz_ratio"] = safe_divide(
    df_model["i_accommodation_food_biz"], df_model["biz_cnt"]
)

df_model["retail_biz_ratio"] = safe_divide(
    df_model["g_wholesale_retail_biz"], df_model["biz_cnt"]
)

df_model["education_biz_ratio"] = safe_divide(
    df_model["p_education_biz"], df_model["biz_cnt"]
)

df_model["culture_biz_ratio"] = safe_divide(
    df_model["r_arts_sports_biz"], df_model["biz_cnt"]
)

df_model["info_comm_biz_ratio"] = safe_divide(
    df_model["j_info_comm_biz"], df_model["biz_cnt"]
)

# 상권 밀도
df_model["store_per_area"] = safe_divide(
    df_model["store_count"], df_model["commercial_area_count"]
)

# =========================
# 5. KMeans Feature 선택
# =========================

cluster_features = [
    # 인구 특성
    "young_pop_ratio",
    "middle_old_pop_ratio",
    # 오피스 / 직장인 특성
    "office_worker_ratio",
    # 교통 접근성
    "subway_total_psngr_co",
    "subway_station_count",
    # 상권 규모
    "commercial_area_count",
    "store_count",
    "store_per_area",
    # 업종 특성
    "food_biz_ratio",
    "retail_biz_ratio",
    "education_biz_ratio",
    "culture_biz_ratio",
    "info_comm_biz_ratio",
    # 점포 경쟁 / 안정성
    "similar_store_ratio",
    "franchise_ratio",
    "open_store_ratio",
    "close_store_ratio",
    "net_open_store_ratio",
    "store_turnover_ratio",
]

X = df_model[cluster_features].copy()
X = X.replace([np.inf, -np.inf], np.nan).fillna(0)

print("사용 feature 수:", len(cluster_features))
display(X.head())

# =========================
# 6. 스케일링
# =========================

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# =========================
# 7. k값 비교
# =========================

k_results = []

for k in range(2, min(8, len(df_model))):
    temp_kmeans = KMeans(n_clusters=k, random_state=42, n_init=20)

    temp_labels = temp_kmeans.fit_predict(X_scaled)

    k_results.append(
        {
            "k": k,
            "inertia": temp_kmeans.inertia_,
            "silhouette_score": silhouette_score(X_scaled, temp_labels),
            "davies_bouldin_score": davies_bouldin_score(X_scaled, temp_labels),
        }
    )

k_eval = pd.DataFrame(k_results)

print("k값 비교 결과")
display(k_eval)

# =========================
# 8. KMeans 실행
# =========================

kmeans = KMeans(n_clusters=K, random_state=42, n_init=20)

cluster_labels = kmeans.fit_predict(X_scaled)

df_model["cluster_id"] = cluster_labels

# =========================
# 9. PCA 2차원 축소
# =========================

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

df_model["pca_x"] = X_pca[:, 0]
df_model["pca_y"] = X_pca[:, 1]

# =========================
# 10. 결과 테이블 생성
# =========================

result_cols = [
    "dong_nm",
    "cluster_id",
    "pca_x",
    "pca_y",
    "biz_cnt",
    "emp_cnt",
    "tot_pop",
    "pop_20s",
    "pop_30s",
    "subway_total_psngr_co",
    "subway_station_count",
    "commercial_area_count",
    "store_count",
    "similar_store_ratio",
    "franchise_ratio",
    "open_store_ratio",
    "close_store_ratio",
    "net_open_store_ratio",
    "store_turnover_ratio",
] + cluster_features

# 중복 제거
result_cols = list(dict.fromkeys(result_cols))

dong_cluster_result = df_model[result_cols].copy()

# =========================
# 11. 군집별 평균값 생성
# =========================

cluster_summary = (
    dong_cluster_result.groupby("cluster_id")[cluster_features].mean().reset_index()
)

# =========================
# 12. 군집명 자동 초안 생성
# =========================

summary = cluster_summary.copy()

score_cols = [
    "young_pop_ratio",
    "middle_old_pop_ratio",
    "office_worker_ratio",
    "subway_total_psngr_co",
    "store_count",
    "food_biz_ratio",
    "education_biz_ratio",
    "franchise_ratio",
    "store_turnover_ratio",
]

for col in score_cols:
    summary[f"{col}_s"] = minmax_series(summary[col])

cluster_name_map = {}

for _, row in summary.iterrows():
    cluster_id = int(row["cluster_id"])

    type_scores = {
        "오피스형 상권": (
            row["office_worker_ratio_s"] * 0.55
            + row["subway_total_psngr_co_s"] * 0.25
            + row["store_count_s"] * 0.20
        ),
        "대학가/청년형 상권": (
            row["young_pop_ratio_s"] * 0.50
            + row["education_biz_ratio_s"] * 0.30
            + row["food_biz_ratio_s"] * 0.20
        ),
        "관광/트렌드형 상권": (
            row["subway_total_psngr_co_s"] * 0.30
            + row["store_count_s"] * 0.30
            + row["food_biz_ratio_s"] * 0.25
            + row["franchise_ratio_s"] * 0.15
        ),
        "주거/생활밀착형 상권": (
            row["middle_old_pop_ratio_s"] * 0.35
            + (1 - row["subway_total_psngr_co_s"]) * 0.20
            + (1 - row["store_turnover_ratio_s"]) * 0.20
            + row["franchise_ratio_s"] * 0.15
            + row["store_count_s"] * 0.10
        ),
    }

    cluster_name = max(type_scores, key=type_scores.get)
    cluster_name_map[cluster_id] = cluster_name

dong_cluster_result["cluster_name"] = dong_cluster_result["cluster_id"].map(
    cluster_name_map
)
cluster_summary["cluster_name"] = cluster_summary["cluster_id"].map(cluster_name_map)

# =========================
# 13. 평가 지표
# =========================

metrics = {
    "n_clusters": K,
    "n_rows": len(df_model),
    "n_features": len(cluster_features),
    "inertia": float(kmeans.inertia_),
    "silhouette_score": float(silhouette_score(X_scaled, cluster_labels)),
    "davies_bouldin_score": float(davies_bouldin_score(X_scaled, cluster_labels)),
    "pca_explained_variance_ratio": pca.explained_variance_ratio_.tolist(),
    "feature_cols": cluster_features,
    "cluster_name_map": cluster_name_map,
}

print("KMeans 평가 지표")
print(json.dumps(metrics, ensure_ascii=False, indent=2))

# =========================
# 14. 결과 저장
# =========================

dong_cluster_result.to_csv(
    OUTPUT_DIR / "dong_cluster_result.csv", index=False, encoding="utf-8-sig"
)

cluster_summary.to_csv(
    OUTPUT_DIR / "cluster_summary.csv", index=False, encoding="utf-8-sig"
)

k_eval.to_csv(OUTPUT_DIR / "k_eval.csv", index=False, encoding="utf-8-sig")

df_model[["dong_nm"] + cluster_features].to_csv(
    OUTPUT_DIR / "model_features.csv", index=False, encoding="utf-8-sig"
)

with open(OUTPUT_DIR / "metrics.json", "w", encoding="utf-8") as f:
    json.dump(metrics, f, ensure_ascii=False, indent=2)

joblib.dump(
    {
        "scaler": scaler,
        "kmeans": kmeans,
        "pca": pca,
        "feature_cols": cluster_features,
        "cluster_name_map": cluster_name_map,
    },
    OUTPUT_DIR / "kmeans_pipeline.pkl",
)

# =========================
# 15. 결과 출력
# =========================

print("\n군집별 동 목록")
cluster_dong_list = (
    dong_cluster_result.groupby(["cluster_id", "cluster_name"])["dong_nm"]
    .apply(list)
    .reset_index()
)

display(cluster_dong_list)

print("\n동별 군집 결과")
display(
    dong_cluster_result[
        [
            "dong_nm",
            "cluster_id",
            "cluster_name",
            "young_pop_ratio",
            "office_worker_ratio",
            "subway_total_psngr_co",
            "store_count",
            "similar_store_ratio",
            "franchise_ratio",
            "store_turnover_ratio",
        ]
    ].sort_values(["cluster_id", "dong_nm"])
)

print("\n군집별 평균 feature")
display(cluster_summary)

print("\n군집별 평균 feature - Transpose")
display(cluster_summary.set_index(["cluster_id", "cluster_name"]).T)

# =========================
# 16. PCA 시각화
# =========================
import platform
from matplotlib import font_manager

plt.rcParams["axes.unicode_minus"] = False

# OS별 한글 폰트 설정
os_name = platform.system()  # 'Windows', 'Darwin'(macOS), 'Linux'

if os_name == "Windows":
    preferred_fonts = ["Malgun Gothic"]  # 맑은 고딕
elif os_name == "Darwin":
    preferred_fonts = ["AppleGothic"]  # macOS
else:
    # Linux 대비(선택)
    preferred_fonts = ["NanumGothic", "Noto Sans CJK KR", "DejaVu Sans"]

available = {f.name for f in font_manager.fontManager.ttflist}
selected_font = None

for f in preferred_fonts:
    if f in available:
        selected_font = f
        break

if selected_font:
    plt.rcParams["font.family"] = selected_font
    print(f"[Font] Using: {selected_font} ({os_name})")
else:
    print(f"[Font] Preferred font not found on {os_name}. Using default font.")


plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    dong_cluster_result["pca_x"],
    dong_cluster_result["pca_y"],
    c=dong_cluster_result["cluster_id"],
    s=130,
    alpha=0.85,
)

for _, row in dong_cluster_result.iterrows():
    plt.text(row["pca_x"] + 0.03, row["pca_y"] + 0.03, row["dong_nm"], fontsize=10)

plt.title("마포구 행정동 KMeans 군집화 결과")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.grid(alpha=0.3)
plt.colorbar(scatter, label="cluster_id")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "pca_cluster.png", dpi=160, bbox_inches="tight")

plt.show()

print("\n저장 완료")
print(f"- {OUTPUT_DIR / 'dong_cluster_result.csv'}")
print(f"- {OUTPUT_DIR / 'cluster_summary.csv'}")
print(f"- {OUTPUT_DIR / 'k_eval.csv'}")
print(f"- {OUTPUT_DIR / 'model_features.csv'}")
print(f"- {OUTPUT_DIR / 'metrics.json'}")
print(f"- {OUTPUT_DIR / 'kmeans_pipeline.pkl'}")
print(f"- {OUTPUT_DIR / 'pca_cluster.png'}")
```
