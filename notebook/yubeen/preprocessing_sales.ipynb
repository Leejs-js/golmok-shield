```python
# 매출은 단위가 크고 skew가 매우 커서 log 변환

import numpy as np
import pandas as pd

amt_cols = [col for col in mapo_df.columns if col.endswith("_amt")]

# 로그 컬럼 한 번에 생성
log_df = np.log1p(mapo_df[amt_cols])
log_df.columns = [f"log_{col}" for col in amt_cols]

# 한 번에 붙이기
mapo_df = pd.concat([mapo_df, log_df], axis=1)

mapo_df.head()
```

```python
#상관관계 분석

corr = mapo_df[num_cols].corr()

corr['thsmon_selng_amt'] \
    .sort_values(ascending=False) \
    .to_frame(name="correlation")
```

▼원본 매출 데이터는 규모 효과(Size Effect)가 강하게 반영되어 상권 간 비교가 어려움. 이를 보완하기 위해 주말 매출 비율, 시간대 매출 비율, 연령대 매출 비율, 객단가 등의 파생변수를 생성하는 Feature Engineering을 수행.

```python
drop_cols = [

    # 요일별 세부 매출
    'mon_selng_amt',
    'tue_selng_amt',
    'wed_selng_amt',
    'thu_selng_amt',
    'fri_selng_amt',
    'sat_selng_amt',
    'sun_selng_amt',

    # 요일별 세부 건수
    'mon_selng_co',
    'tue_selng_co',
    'wed_selng_co',
    'thu_selng_co',
    'fri_selng_co',
    'sat_selng_co',
    'sun_selng_co'
]

cluster_df = mapo_df.drop(columns=drop_cols)

#군집화용 핵심 파생변수 생성

cluster_df['weekend_ratio'] = (
    cluster_df['wkend_selng_amt']
    / cluster_df['thsmon_selng_amt']
) #주말 매출 비율

time_cols = [
    'tmzon_00_06_selng_amt',
    'tmzon_06_11_selng_amt',
    'tmzon_11_14_selng_amt',
    'tmzon_14_17_selng_amt',
    'tmzon_17_21_selng_amt',
    'tmzon_21_24_selng_amt'
]

for col in time_cols:
    cluster_df[col + '_ratio'] = (
        cluster_df[col]
        / cluster_df['thsmon_selng_amt']
    ) #시간대 비율

cluster_df['male_ratio'] = (
    cluster_df['ml_selng_amt']
    / cluster_df['thsmon_selng_amt']
) #남성 소비 비율

cluster_df['female_ratio'] = (
    cluster_df['fml_selng_amt']
    / cluster_df['thsmon_selng_amt']
) #여성 소비 비율

age_cols = [
    'age_10_selng_amt',
    'age_20_selng_amt',
    'age_30_selng_amt',
    'age_40_selng_amt',
    'age_50_selng_amt',
    'age_60_up_selng_amt'
]

for col in age_cols:
    cluster_df[col + '_ratio'] = (
        cluster_df[col]
        / cluster_df['thsmon_selng_amt']
    ) #연령대 비율

cluster_df['avg_ticket'] = (
    cluster_df['thsmon_selng_amt']
    / cluster_df['thsmon_selng_co']
) #1 회 방문당 평균 결제 금액

print(cluster_df.shape)

cluster_df.head()
```
