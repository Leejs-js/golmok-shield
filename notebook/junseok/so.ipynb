```python
import pandas as pd
import numpy as np
#2023
# 1.데이터 불러오기
smbd_df = pd.read_csv('소상공인시장진흥공단_상가(상권)정보_서울_202306.csv')
```

```python
#불러온 데이터 확인
print(smbd_df)
```

```python
# =========================================================
# [1단계] 맨 처음: 마포구 데이터만 먼저 필터링 (가장 강력추천!)
# =========================================================
# 시군구코드가 11440 이거나, 시군구명이 '마포구'인 데이터만 필터링
smbd_mapo_raw_df = smbd_df[(smbd_df['시군구코드'] == 11440) | (smbd_df['시군구명'] == '마포구')].copy()

print(f"1단계 완료: 마포구 데이터만 {len(smbd_mapo_raw_df)}건 추출 완료 (용량 최적화!)")
print(smbd_mapo_raw_df)
```

```python
# =========================================================
# [2단계] 1~3순위 핵심 컬럼 7개만 추출 및 데이터 정제
# =========================================================
# 1순위(기준): 시군구명, 행정동명, 상권업종중분류명
# 2순위(지도): 경도, 위도
# 3순위(리포트): 건물명, 상호명

essential_cols = ['시군구명', '행정동명', '상권업종중분류명', '경도', '위도', '건물명', '상호명']
smbd_df_clean = smbd_mapo_raw_df[essential_cols].copy()

# 텍스트 컬럼들만 안전하게 앞뒤 공백 제거 (숫자형인 경도/위도 에러 방지)
string_cols = ['시군구명', '행정동명', '상권업종중분류명', '건물명', '상호명']
for col in string_cols:
    smbd_df_clean[col] = smbd_df_clean[col].astype(str).str.strip()
    
    
print(smbd_df_clean)
```

```python
# 빈 칸("")이나 문자열 'nan'으로 표기된 불량 데이터를 진짜 결측치(NaN)로 통일
smbd_df_clean = smbd_df_clean.replace({'nan': np.nan, '': np.nan})
print(smbd_df_clean)
```

```python
# 3순위 컬럼(건물명, 상호명)의 널값은 삭제하지 않고 "정보없음"으로 안전하게 채우기
smbd_df_clean['상호명'] = smbd_df_clean['상호명'].fillna("정보없음")
smbd_df_clean['건물명'] = smbd_df_clean['건물명'].fillna("정보없음")
print(smbd_df_clean)
```

```python
# 1~2순위 핵심 필수 컬럼에 결측치가 있는 행만 과감하게 제거
smbd_df_clean = smbd_df_clean.dropna(subset=['시군구명', '행정동명', '상권업종중분류명', '경도', '위도'])

print(smbd_df_clean)
```

```python
# # =========================================================
# # 💡 [무기 1] 나중에 지도 시각화 및 최종 솔루션 팝업에 쓸 마스터 데이터
# # =========================================================
smbd_df_mapo_gis = smbd_df_clean.copy()
print(f"✅ 지도용 마스터 데이터 생성 완료 (점포 수: {len(smbd_df_mapo_gis)}개, 컬럼 수: {len(smbd_df_mapo_gis.columns)}개)")
```

```python
# =========================================================
# [3단계] 머신러닝용 테이블 정리 (행정동별 업종 밀집도 계산)
# =========================================================
# 가로(Column)는 업종명, 세로(Row)는 마포구 행정동명으로 묶어서 점포 개수 피벗
smbd_pivot_mapo = pd.crosstab(smbd_df_clean['행정동명'], smbd_df_clean['상권업종중분류명'])
print(smbd_df_clean)
```
