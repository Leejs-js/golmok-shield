### merged_datamart와 merged_상권 결합

```python
datamart_path = PREPROCESSED_DIR / "merged_datamart.csv"
store_path = PREPROCESSED_DIR / "merged_상권.csv"

df_datamart = read_csv_auto(datamart_path)
df_store = read_csv_auto(store_path)
```

```python
# =========================
# 1. 병합 전 기본 확인
# =========================

print("df_datamart shape:", df_datamart.shape)
print("df_store shape:", df_store.shape)

print("\n[df_datamart dong_nm 개수]")
print(df_datamart["dong_nm"].nunique())

print("\n[df_store dong_nm 개수]")
print(df_store["dong_nm"].nunique())


# =========================
# 2. dong_nm 정리
# =========================
# 공백 때문에 매칭 안 되는 경우 방지

df_datamart["dong_nm"] = df_datamart["dong_nm"].astype(str).str.strip()
df_store["dong_nm"] = df_store["dong_nm"].astype(str).str.strip()


# =========================
# 3. 병합 전 동 목록 비교
# =========================

datamart_dongs = set(df_datamart["dong_nm"].unique())
store_dongs = set(df_store["dong_nm"].unique())

only_datamart = sorted(datamart_dongs - store_dongs)
only_store = sorted(store_dongs - datamart_dongs)
common_dongs = sorted(datamart_dongs & store_dongs)

print("\n[datamart에만 있는 동]")
print(only_datamart)

print("\n[store에만 있는 동]")
print(only_store)

print("\n[공통 동 개수]")
print(len(common_dongs))


# =========================
# 4. store 쪽 dong_nm 중복 확인
# =========================
# df_store는 dong_nm 기준으로 1행이어야 안전함.
# 중복이면 merge 후 행이 늘어남.

store_dup = (
    df_store["dong_nm"]
    .value_counts()
    .reset_index()
)

store_dup.columns = ["dong_nm", "count"]
store_dup = store_dup[store_dup["count"] > 1]

print("\n[df_store dong_nm 중복]")
display(store_dup)

if len(store_dup) > 0:
    print("주의: df_store에 dong_nm 중복이 있어서 merge 후 행 수가 늘어날 수 있습니다.")


# =========================
# 5. indicator로 병합 상태 확인
# =========================

df_merged_check = pd.merge(
    df_datamart,
    df_store,
    on="dong_nm",
    how="left",
    indicator=True
)

print("\n[merge 결과 shape]")
print(df_merged_check.shape)

print("\n[merge 상태]")
print(df_merged_check["_merge"].value_counts())


# =========================
# 6. 매칭 실패한 행 확인
# =========================

unmatched_rows = df_merged_check[
    df_merged_check["_merge"] == "left_only"
].copy()

print("\n[매칭 실패 행 수]")
print(len(unmatched_rows))

print("\n[매칭 실패 dong_nm]")
print(sorted(unmatched_rows["dong_nm"].unique()))

display(unmatched_rows.head())


# =========================
# 7. 병합하면서 새로 붙은 컬럼 결측 확인
# =========================

store_cols = [
    col for col in df_store.columns
    if col != "dong_nm"
]

missing_after_merge = (
    df_merged_check[store_cols]
    .isna()
    .sum()
    .reset_index()
)

missing_after_merge.columns = ["column", "missing_count"]
missing_after_merge["missing_ratio"] = (
    missing_after_merge["missing_count"] / len(df_merged_check)
)

missing_after_merge = missing_after_merge.sort_values(
    "missing_count",
    ascending=False
)

print("\n[store에서 붙은 컬럼의 결측 현황]")
display(missing_after_merge)


# =========================
# 8. 최종 병합본 생성
# =========================

df_merged = df_merged_check.drop(columns=["_merge"])

print("\n[최종 df_merged shape]")
print(df_merged.shape)

display(df_merged.head())
```

```python
save_path = PREPROCESSED_DIR / "merged_datamart.csv"

df_merged.to_csv(save_path, index=False, encoding="utf-8-sig")

save_path
```
