```python
import pandas as pd
import numpy as np
# 1. 아까 저장했던 지도용 마스터 데이터 파일 불러오기
df_gis = pd.read_csv('마포구_점포_마스터_지도용.csv')

# 2. [안전 장치] 시군구명이 '마포구'인 데이터만 다시 한번 확실하게 필터링
# (이미 마포구만 저장되어 있어서 행 개수 변화가 거의 없을 것입니다!)
smbd_df_mapo_gis = df_gis[df_gis['시군구명'] == '마포구'].copy()

# 3. 잘 추출되었는지 검증 확인
print("=========================================================")
print(f"🗺️ 지도용 마스터 내 마포구 점포 수: {len(smbd_df_mapo_gis)}개")
print("---------------------------------------------------------")
print("포함된 행정동 목록:")
print(smbd_df_mapo_gis['행정동명'].unique())
print("=========================================================")

# 2. 지도 시각화 및 리포트용 마스터 데이터 저장
# (경도, 위도, 상호명, 건물명이 포함된 데이터입니다)
smbd_df_mapo_gis.to_csv('마포구_점포_마스터_지도용.csv', index=False, encoding='utf-8-sig')

# 이제 이 df_mapo_only 데이터프레임을 위도/경도 기반 지도 시각화에 그대로 사용하시면 됩니다!
```
